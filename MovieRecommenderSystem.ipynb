```python
#inside the datasets there are 9 columns,but we do not need all of them
movies.columns
```

```python
#feature selection-since we do not require all columns we will select some of them(important ones)
"""id-moviesid,
title-movie names
overview-what the story/movie is all about
genre-type of movies(crime,comedy,historical,romantic,horror...etc)"""
movies=movies[['id','title','overview','genre']]
movies
```

```python
#one extra column added "tags" that contents the combined overall overviws + genre
movies['tags']=movies['overview']+movies['genre']
movies
```

```python
"""since we had already combined column overview & genre we do not need their individual columns
so we drop them"""
new_data=movies.drop(columns=['overview','genre'])
new_data
```

```python
#max_features because there are 10000 movies(rows) & stop_words= default english we are taking
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=10000,stop_words='english')
cv
```

```python
# now you need to transform the dataset into vector form
vector=cv.fit_transform(new_data['tags'].values.astype('U')).toarray()
vector.shape  # 10000 columns and 10000 rows
```

```python
"""now we need to built a recommender systems that will use Cosine Similarity. if you search avatars movie its an action movie
so you expect that your system will recommend you similar action movies so it returns ironman."""
from sklearn.metrics.pairwise import cosine_similarity
similarity=cosine_similarity(vector)
similarity
```

```python
# trying to fetch the index number to uniquely identify titles
new_data[new_data['title']=="The Godfather"].index[0]
```

```python
"""Take all similarity scores for movie #2, pair each score with its movie index(enumerate),
and sort them in descending order so the most similar movies come first"""
distance=sorted(list(enumerate(similarity[2])), reverse=True,key=lambda vector:vector[1])
for i in distance[0:5]: # will return top 5 movies most similar to index 2 movie
  print(new_data.iloc[i[0]].title)
```

```python
# The previous logic works only for indexed 2 movie but this logic is going to work for all
def recommend(movie_title, top_n=5):
    # Case-insensitive search for the movie
    matches = new_data[new_data['title'].str.lower() == movie_title.lower()]
    if matches.empty:
        return ["Movie not found in dataset."]

    # Get index of the movie
    index = matches.index[0]

    # Calculate similarity scores
    distances = sorted(list(enumerate(similarity[index])), key=lambda x: x[1], reverse=True)

    # Get top N similar movies (excluding the movie itself)
    movie_indices = [i[0] for i in distances[1:top_n+1]]

    return new_data.iloc[movie_indices]['title'].tolist()
```

```python
print(recommend("iron man"))
```
